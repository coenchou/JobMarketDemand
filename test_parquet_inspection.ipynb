# Parquet Metadata & Encoding Inspection

This notebook is a lightweight test for understanding how the current `.parquet` files are encoded and whether `company_id`, `skill_id`, and `r0_id` are simple numeric labels or pre-encoded values.

```python
import pandas as pd
import pyarrow.parquet as pq
from pathlib import Path

datasets_dir = Path('datasets')
company_path = datasets_dir / 'company.parquet'
region_path = datasets_dir / 'region.parquet'
r0_path = datasets_dir / 'r0.parquet'

print('=== PARQUET METADATA INSPECTION ===')
for path, label in [(company_path, 'company'), (region_path, 'region'), (r0_path, 'r0')]:
    pf = pq.ParquetFile(path)
    print(f"\n{label}.parquet")
    print(f"  rows: {pf.metadata.num_rows}")
    print(f"  columns: {len(pf.schema.names)}")
    print(f"  schema: {pf.schema}")
    print(f"  row groups: {pf.num_row_groups}")
    print('  compression by column:')
    row_group = pf.metadata.row_group(0)
    for i in range(row_group.num_columns):
        col = row_group.column(i)
        print(f"    {col.path_in_schema}: {col.compression}")
        print(f"      encodings: {list(col.encodings)}")

print('\n=== CATEGORICAL / ENCODING HINTS ===')
datasets = [
    (pd.read_parquet(company_path, engine='fastparquet'), 'company', ['company_id', 'skill_id']),
    (pd.read_parquet(region_path, engine='fastparquet'), 'region', ['company_id', 'skill_id']),
    (pd.read_parquet(r0_path, engine='fastparquet'), 'r0', ['r0_id', 'skill_id']),
]
for df, label, cat_cols in datasets:
    print(f"\n{label} dataset")
    for col in cat_cols:
        if col in df.columns:
            unique_count = df[col].nunique(dropna=False)
            sample = df[col].dropna().unique()[:10]
            print(f"  {col}: unique values = {unique_count}, sample = {sample}")

print('\n=== QUICK VALUE CHECK ===')
for label, df in [(d[1], d[0]) for d in datasets]:
    print(f"\n{label} dataset values")
    for col in ['company_id', 'skill_id', 'r0_id']:
        if col in df.columns:
            print(f"  {label}[{col}] head: {df[col].head(10).tolist()}")
```
