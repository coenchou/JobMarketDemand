# Parser Testing Notebook

This notebook runs the resume parser on each sample resume in its own cell so you can inspect the output one file at a time.

The parser script is expected to be available at `src/parser.py` and the resumes are under `resumes/`.


```python
import json
from pathlib import Path

from src.parser import parse_resume_file

ROOT = Path.cwd()
RESUME_DIR = ROOT / 'resumes'
OUTPUT_DIR = ROOT / 'outputs' / 'parser'
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print('ROOT:', ROOT)
print('RESUME_DIR:', RESUME_DIR)
print('OUTPUT_DIR:', OUTPUT_DIR)
```

```python
def run_parser_for_resume(resume_path: str):
    path = Path(resume_path)
    if not path.exists():
        raise FileNotFoundError(f'Resume not found: {path}')

    result = parse_resume_file(str(path))
    output_path = OUTPUT_DIR / f"{path.stem}.json"
    output_path.write_text(json.dumps(result, indent=2, ensure_ascii=False), encoding='utf-8')
    print(f'Parsed {path.name}')
    print(f'Wrote {output_path}')
    return result
```

```python
resume_files = sorted(RESUME_DIR.glob('*.txt'))
resume_files
```

```python
resume_1 = resume_files[0]
resume_1
```

```python
result_1 = run_parser_for_resume(resume_1)
result_1
```

```python
resume_2 = resume_files[1]
resume_2
```

```python
result_2 = run_parser_for_resume(resume_2)
result_2
```

```python
resume_3 = resume_files[2]
resume_3
```

```python
result_3 = run_parser_for_resume(resume_3)
result_3
```

```python
for resume_path in resume_files:
    print(f'--- {resume_path.name} ---')
    try:
        parsed = run_parser_for_resume(str(resume_path))
        print('skills:', parsed.get('skills', [])[:10])
        print('education:', parsed.get('education', []))
        print('soc_candidates:', parsed.get('soc_candidates', []))
    except Exception as exc:
        print(f'ERROR for {resume_path.name}: {exc}')
```

```python
for resume_path in resume_files:
    print(f'Checking {resume_path.name}')
    print('exists:', resume_path.exists())
    print('size:', resume_path.stat().st_size if resume_path.exists() else 'n/a')
```
